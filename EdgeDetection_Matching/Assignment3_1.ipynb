```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the image
img_color = cv2.imread('Assignment3_Files/Credit.png')

# 2. Convert to grayscale
img_gray = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
plt.figure(figsize=(5,5))
plt.title("Grayscale Image")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# Sobel with dx=1, dy=0, derivative in X-direction
sobel_x_8u = cv2.Sobel(img_gray, cv2.CV_8U, 1, 0, ksize=3)
sobel_x_64f = cv2.Sobel(img_gray, cv2.CV_64F, 1, 0, ksize=3)

# To visualize the magnitude of the 64F result properly, take the absolute value and convert to 8-bit:
sobel_x_64f_abs = np.absolute(sobel_x_64f)
sobel_x_64f_8u  = np.uint8(sobel_x_64f_abs)

# Display
plt.figure(figsize=(10,5))
plt.subplot(1,2,1), plt.imshow(sobel_x_8u, cmap='gray'), plt.title("Sobel X CV_8U")
plt.axis('off')
plt.subplot(1,2,2), plt.imshow(sobel_x_64f_8u, cmap='gray'), plt.title("Sobel X CV_64F")
plt.axis('off')
plt.show()
```

```python
# Sobel with dx=0, dy=1
sobel_y_8u = cv2.Sobel(img_gray, cv2.CV_8U, 0, 1, ksize=3)
sobel_y_64f = cv2.Sobel(img_gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_y_64f_abs = np.absolute(sobel_y_64f)
sobel_y_64f_8u  = np.uint8(sobel_y_64f_abs)

# Display
plt.figure(figsize=(10,5))
plt.subplot(1,2,1), plt.imshow(sobel_y_8u, cmap='gray'), plt.title("Sobel Y CV_8U")
plt.axis('off')
plt.subplot(1,2,2), plt.imshow(sobel_y_64f_8u, cmap='gray'), plt.title("Sobel Y CV_64F")
plt.axis('off')
plt.show()
```

```python
# Sobel with dx=1, dy=1
sobel_xy_8u = cv2.Sobel(img_gray, cv2.CV_8U, 1, 1, ksize=3)
sobel_xy_64f = cv2.Sobel(img_gray, cv2.CV_64F, 1, 1, ksize=3)
sobel_xy_64f_abs = np.absolute(sobel_xy_64f)
sobel_xy_64f_8u  = np.uint8(sobel_xy_64f_abs)

# Display
plt.figure(figsize=(10,5))
plt.subplot(1,2,1), plt.imshow(sobel_xy_8u, cmap='gray'), plt.title("Sobel X+Y CV_8U")
plt.axis('off')
plt.subplot(1,2,2), plt.imshow(sobel_xy_64f_8u, cmap='gray'), plt.title("Sobel X+Y CV_64F")
plt.axis('off')
plt.show()
```

```python
# Different thresholds for Canny
edges_canny_1 = cv2.Canny(img_gray, threshold1=50, threshold2=150)
edges_canny_2 = cv2.Canny(img_gray, threshold1=100, threshold2=200)

# Display the results
plt.figure(figsize=(10,5))
plt.subplot(1,2,1), plt.imshow(edges_canny_1, cmap='gray'), plt.title("Canny (50,150)")
plt.axis('off')
plt.subplot(1,2,2), plt.imshow(edges_canny_2, cmap='gray'), plt.title("Canny (100,200)")
plt.axis('off')
plt.show()
```
